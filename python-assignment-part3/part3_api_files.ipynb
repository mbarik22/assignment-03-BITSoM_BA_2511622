Task 1 — File Read & Write Basics

```python
# Task 1 — File Read & Write Basics

file_name = "python_notes.txt"

# ----------------------------
# Part A.1 — Write to file
# ----------------------------
try:
    with open(file_name, "w", encoding="utf-8") as file:
        file.write("Topic 1: Variables store data. Python is dynamically typed.\n")
        file.write("Topic 2: Lists are ordered and mutable.\n")
        file.write("Topic 3: Dictionaries store key-value pairs.\n")
        file.write("Topic 4: Loops automate repetitive tasks.\n")
        file.write("Topic 5: Exception handling prevents crashes.\n")
    
    print("File written successfully.")

except Exception as e:
    print(f"Error while writing to file: {e}")


# ----------------------------
# Part A.2 — Append to file
# ----------------------------
try:
    with open(file_name, "a", encoding="utf-8") as file:
        file.write("Topic 6: Functions help reuse code efficiently.\n")
        file.write("Topic 7: Modules help organize large programs.\n")
    
    print("Lines appended successfully.")

except Exception as e:
    print(f"Error while appending to file: {e}")
```

```python
# ----------------------------
# Part B — Read from file
# ----------------------------

try:
    with open(file_name, "r", encoding="utf-8") as file:
        lines = file.readlines()

    # 1. Print numbered lines (without \n)
    print("\n--- File Contents ---")
    for idx, line in enumerate(lines, start=1):
        print(f"{idx}. {line.strip()}")

    # 2. Count total lines
    total_lines = len(lines)
    print(f"\nTotal number of lines: {total_lines}")

    # 3. Keyword search (case-insensitive)
    keyword = input("\nEnter a keyword to search: ").strip().lower()

    matches = [line.strip() for line in lines if keyword in line.lower()]

    if matches:
        print("\nMatching lines:")
        for match in matches:
            print(f"- {match}")
    else:
        print("\nNo matching lines found. Try a different keyword.")

except FileNotFoundError:
    print("Error: File not found. Please run Part A first to create the file.")

except Exception as e:
    print(f"An error occurred while reading the file: {e}")
```

Task 2 — API Integration


Step 1 — Fetch and Display Products:

```python
import requests

BASE_URL = "https://dummyjson.com/products?limit=20"

try:
    # Make API request
    response = requests.get(BASE_URL, timeout=10)
    response.raise_for_status()  # Raises HTTPError for bad responses

    data = response.json()

    products = data.get("products", [])

    if not products:
        print("No products found in API response.")
    else:
        # Print table header
        print(f"{'ID':<4} | {'Title':<30} | {'Category':<15} | {'Price':<10} | {'Rating'}")
        print("-" * 80)

        # Print each product row
        for product in products:
            pid = product.get("id", "N/A")
            title = product.get("title", "N/A")[:30]  # truncate long titles
            category = product.get("category", "N/A")
            price = f"${product.get('price', 0):.2f}"
            rating = product.get("rating", "N/A")

            print(f"{pid:<4} | {title:<30} | {category:<15} | {price:<10} | {rating}")

except requests.exceptions.Timeout:
    print("Error: Request timed out. Please try again later.")

except requests.exceptions.ConnectionError:
    print("Error: Failed to connect to the API. Check your internet connection.")

except requests.exceptions.HTTPError as e:
    print(f"HTTP Error occurred: {e}")

except ValueError:
    print("Error: Failed to parse JSON response.")

except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

Step 2 — Filter and Sort:

```python
try:
    # 1. Filter products with rating ≥ 4.5
    filtered_products = [
        p for p in products if p.get("rating", 0) >= 4.5
    ]

    # 2. Sort by price (descending)
    sorted_products = sorted(
        filtered_products,
        key=lambda x: x.get("price", 0),
        reverse=True
    )

    # 3. Print result
    if not sorted_products:
        print("No products found with rating ≥ 4.5")
    else:
        print(f"{'ID':<4} | {'Title':<30} | {'Price':<10} | {'Rating'}")
        print("-" * 65)

        for p in sorted_products:
            print(
                f"{p.get('id', 'N/A'):<4} | "
                f"{p.get('title', 'N/A')[:30]:<30} | "
                f"${p.get('price', 0):<10.2f} | "
                f"{p.get('rating', 'N/A')}"
            )

except requests.exceptions.RequestException as e:
    print(f"API request failed: {e}")

except Exception as e:
    print(f"Unexpected error: {e}")
```

Task 3 — Exception Handling

Part A — Guarded Calculator:

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid input types"


# Test cases
print(safe_divide(10, 2))       # Expected: 5.0
print(safe_divide(10, 0))       # Expected: Error message
print(safe_divide("ten", 2))    # Expected: Error message
```

Part B — Guarded File Reader:

```python
def read_file_safe(filename):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            content = file.read()
            return content

    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")

    finally:
        print("File operation attempt complete.")


# Test cases
print("\n--- Reading existing file ---")
result1 = read_file_safe("python_notes.txt")
if result1:
    print(result1)

print("\n--- Reading missing file ---")
result2 = read_file_safe("ghost_file.txt")
if result2:
    print(result2)
```

Part C — Robust API Calls:

```python
import requests

# ----------------------------
# Safe GET Request
# ----------------------------
def safe_get(url):
    try:
        response = requests.get(url, timeout=5)

        # Handle HTTP errors (like 404, 500)
        if response.status_code != 200:
            print(f"HTTP Error: {response.status_code}")
            return None

        return response.json()

    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")

    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")

    except Exception as e:
        print(f"Unexpected error: {e}")

    return None


# ----------------------------
# Safe POST Request
# ----------------------------
def safe_post(url, payload):
    try:
        response = requests.post(url, json=payload, timeout=5)

        if response.status_code not in (200, 201):
            print(f"HTTP Error: {response.status_code}")
            return None

        return response.json()

    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")

    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")

    except Exception as e:
        print(f"Unexpected error: {e}")

    return None

##########################################
#     Using It in Your Existing Tasks    #
##########################################
url = "https://dummyjson.com/products?limit=20"

data = safe_get(url)

if data:
    for product in data.get("products", []):
        print(product["title"], "-", product["price"])

print("====================================\n")
url = "https://dummyjson.com/products/category/laptops"

data = safe_get(url)

if data:
    for product in data.get("products", []):
        print(product["title"], "-", product["price"])

print("====================================\n")
url = "https://dummyjson.com/products/add"

payload = {
    "title": "My Custom Product",
    "price": 999,
    "category": "electronics",
    "description": "A product I created via API"
}

response = safe_post(url, payload)

if response:
    print(response)        
```

Part D — Input Validation Loop:

```python
import requests

BASE_URL = "https://dummyjson.com/products/"

while True:
    user_input = input("\nEnter a product ID to look up (1-100), or 'quit' to exit: ").strip()

    # Exit condition
    if user_input.lower() == "quit":
        print("Exiting program.")
        break

    # Validate integer input
    if not user_input.isdigit():
        print("Invalid input. Please enter a number between 1 and 100.")
        continue

    product_id = int(user_input)

    # Validate range
    if product_id < 1 or product_id > 100:
        print("Invalid range. Please enter a number between 1 and 100.")
        continue

    # Make API call
    try:
        response = requests.get(BASE_URL + str(product_id), timeout=5)

        # Handle HTTP responses
        if response.status_code == 404:
            print("Product not found.")

        elif response.status_code == 200:
            product = response.json()
            print(f"Title: {product.get('title', 'N/A')}")
            print(f"Price: ${product.get('price', 0):.2f}")

        else:
            print(f"Unexpected HTTP status: {response.status_code}")

    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")

    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")

    except Exception as e:
        print(f"Unexpected error: {e}")
```

Task 4 — Logging to File

```python
import requests
from datetime import datetime

LOG_FILE = "error_log.txt"


# ----------------------------
# Logger Function
# ----------------------------
def log_error(context, error_type, message):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    log_entry = f"[{timestamp}] ERROR in {context}: {error_type} — {message}\n"

    try:
        with open(LOG_FILE, "a", encoding="utf-8") as file:
            file.write(log_entry)
    except Exception as e:
        print(f"Failed to write log: {e}")


# ----------------------------
# 1. Trigger ConnectionError
# ----------------------------
try:
    bad_url = "https://this-host-does-not-exist-xyz.com/api"
    requests.get(bad_url, timeout=5)

except requests.exceptions.ConnectionError as e:
    log_error("fetch_products", "ConnectionError", str(e))


# ----------------------------
# 2. Trigger HTTP Error (404)
# ----------------------------
try:
    url = "https://dummyjson.com/products/999"
    response = requests.get(url, timeout=5)

    if response.status_code != 200:
        log_error(
            "lookup_product",
            "HTTPError",
            f"{response.status_code} Not Found for product ID 999"
        )

except requests.exceptions.ConnectionError:
    log_error("lookup_product", "ConnectionError", "Network failure")

except requests.exceptions.Timeout:
    log_error("lookup_product", "Timeout", "Request timed out")

except Exception as e:
    log_error("lookup_product", "UnexpectedError", str(e))


# ----------------------------
# 3. Read and Print Log File
# ----------------------------
print("\n--- Error Log Contents ---")

try:
    with open(LOG_FILE, "r", encoding="utf-8") as file:
        print(file.read())

except FileNotFoundError:
    print("No log file found.")
```
