#Task 2 — Cart Operations 

```python
cart = []

def add_to_cart(item_name, quantity):
    # Check if item exists
    if item_name not in menu:
        print(f"Item '{item_name}' not found in menu.")
        return
    
    # Check availability
    if not menu[item_name]["available"]:
        print(f"'{item_name}' is currently unavailable.")
        return
    
    # Check if already in cart
    for entry in cart:
        if entry["item"] == item_name:
            entry["quantity"] += quantity
            print(f"Updated '{item_name}' quantity to {entry['quantity']}.")
            return
    
    # Add new item
    cart.append({
        "item": item_name,
        "quantity": quantity,
        "price": menu[item_name]["price"]
    })
    print(f" Added '{item_name}' (x{quantity}) to cart.")
```

Write logic to remove an item from the cart by name.

If the item is not in the cart, print a message.

```python
def remove_from_cart(item_name):
    for i, entry in enumerate(cart):
        if entry["item"] == item_name:
            cart.pop(i)
            print(f" Removed '{item_name}' from cart.")
            return
    
    # If item not found
    print(f" Item '{item_name}' not found in cart!")

```

Write logic to update the quantity of an item already in the cart.

```python
def update_quantity(item_name, new_qty):
    for entry in cart:
        if entry["item"] == item_name:
            
            # If quantity is 0 or less → remove item
            if new_qty <= 0:
                cart.remove(entry)
                print(f" '{item_name}' removed from cart (quantity set to {new_qty}).")
            else:
                entry["quantity"] = new_qty
                print(f" Updated '{item_name}' quantity to {new_qty}.")
            return
    
    # If item not found
    print(f" Item '{item_name}' not found in cart.")
```

Simulate order fulfilment: deduct the quantities from the final cart in Task 2 from the corresponding items in inventory.

If stock for an item is insufficient, print a warning and deduct only what is available (do not go below 0).

```python
def fulfill_order(cart, inventory):
    print("\n Order Fulfilment:\n")
    
    for item in cart:
        name = item["item"]
        required_qty = item["quantity"]
        
        # Check if item exists in inventory
        if name not in inventory:
            print(f" '{name}' not found in inventory!")
            continue
        
        available_stock = inventory[name]["stock"]
        
        if available_stock >= required_qty:
            # Enough stock
            inventory[name]["stock"] -= required_qty
            print(f" {name}: Deducted {required_qty}, Remaining stock: {inventory[name]['stock']}")
        else:
            # Not enough stock
            print(f" {name}: Only {available_stock} available, cannot fulfill {required_qty}")
            inventory[name]["stock"] = 0
            print(f" Deducted {available_stock}, Remaining stock: 0")

fulfill_order(cart, inventory)

print("\n Updated Inventory (sample):")
for name in cart:
    item_name = name["item"]
    print(item_name, "->", inventory[item_name])
```

After deduction, loop through inventory and print a Reorder Alert for every item whose stock is at or below its reorder_level:

```python
def check_reorder_alerts(inventory):
    print("\n Reorder Alerts:\n")
    
    alert_found = False
    
    for item_name, details in inventory.items():
        stock = details["stock"]
        reorder_level = details["reorder_level"]
        
        if stock <= reorder_level:
            print(f"⚠ Reorder Alert: {item_name} — Only {stock} unit(s) left (reorder level: {reorder_level})")
            alert_found = True
    
    if not alert_found:
        print(" All items are sufficiently stocked.")


# --- Run after order fulfilment ---
check_reorder_alerts(inventory)
```

Task 4 — Daily Sales Log Analysis 

```python
# --- Function to compute stats ---
def analyze_sales(sales_log):
    print("\n Revenue Per Day:")
    
    revenue_per_day = {}
    
    for date, orders in sales_log.items():
        total = sum(order["total"] for order in orders)
        revenue_per_day[date] = total
        print(f"{date} → ₹{total:.2f}")
    
    # Best-selling day
    best_day = max(revenue_per_day, key=revenue_per_day.get)
    print(f"\n Best-Selling Day: {best_day} (₹{revenue_per_day[best_day]:.2f})")
    
    # Most ordered item (by number of orders it appears in)
    item_count = {}
    
    for orders in sales_log.values():
        for order in orders:
            for item in set(order["items"]):  # unique items per order
                item_count[item] = item_count.get(item, 0) + 1
    
    most_ordered = max(item_count, key=item_count.get)
    print(f" Most Ordered Item: {most_ordered} ({item_count[most_ordered]} orders)")


# --- Initial Analysis ---
print("-> Initial Analysis:")
print("------------------------------")
analyze_sales(sales_log)


# --- Add new day ---
sales_log["2025-01-05"] = [
    {"order_id": 11, "items": ["Butter Chicken", "Gulab Jamun", "Garlic Naan"], "total": 490.0},
    {"order_id": 12, "items": ["Paneer Tikka", "Rasgulla"],                     "total": 260.0},
]

print("====================================")

# --- Updated Analysis ---
print("\n-> After Adding 2025-01-05:")
print("------------------------------")
analyze_sales(sales_log)

```
